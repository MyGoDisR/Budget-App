```python
import yfinance as yf
import pandas as pd
from datetime import datetime
import requests
from datetime import datetime
from restricted import key

# Text Sentiment analysis
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
def trading_yfin_2(ticker_symbol, start_date, end_date):
  start_date_ = datetime.strptime(start_date, '%Y-%m-%d')
  end_date_ = datetime.strptime(end_date, '%Y-%m-%d')
  ticker = yf.Ticker(ticker_symbol)
  data = ticker.history(start=start_date_, end=end_date_)
  #data.to_csv('lol.csv')
  return data
```

```python
df = trading_yfin_2('AAPL', '2020-01-01', '2025-05-31')
```

```python
import yfinance as yf

dat = yf.Ticker("MSFT")

# get historical market data
dat.history(period='1mo')

# options
dat.option_chain(dat.options[0]).calls

# get financials
dat.balance_sheet
dat.quarterly_income_stmt

# dates
dat.calendar

# general info
dat.info

# analysis
dat.analyst_price_targets

# websocket
dat.live()
```

```python
all_news = dat.news

df_news = pd.DataFrame()

nltk.download('vader_lexicon')
sid = SentimentIntensityAnalyzer()

for i in range(len(all_news)):
    news_ = []
    news_.append(all_news[i]['content']['title'])
    #news_.append(a[i]['content']['provider']['disaplyName'])
    news_.append(all_news[i]['content']['canonicalUrl']['url'])
    news_.append(all_news[i]['content']['pubDate'].split('T')[0])
    news_.append(all_news[i]['content']['summary'])
    sentiment_score = sid.polarity_scores(all_news[i]['content']['summary'])
    news_.append(sentiment_score['compound'])
    # concating news to df
    df_news = pd.concat([df_news, pd.DataFrame([news_], columns=['Title', 'Link', 'Publication Date','Summary','Sentiment'])], ignore_index=True)
    df_news['Feedback'] = df_news['Sentiment'].apply(lambda x: 'good' if x> 0.0 else ('bad' if x < 0.0 else 'neutral'))
```
